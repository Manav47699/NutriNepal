```python
from ultralytics import YOLO
```

```python
# loading the model. 

model = YOLO("../ai-service/models/yolo/best.pt")

# printing classes of the model.
print(model.names) # It detects daal , bhaat and sabji. 
```

```python
# predictions on the image.

results = model.predict(source="images/daal_bhaat.jpeg", save=True, show=True)

for result in results:
    # 1. Bounding Boxes & Classes
    boxes = result.boxes
    for box in boxes:
        cls_id = int(box.cls[0].item())
        conf = float(box.conf[0].item())
        xyxy = box.xyxy[0].tolist()
        print(
            f"Class: {model.names[cls_id]} ({cls_id}) | Conf: {conf:.2f} | Box: {xyxy}"
        )

    # 2. Segmentation Mask Polygons (normalized or pixel coordinates)
    if result.masks is not None:
        # Polygons as list of (N, 2) numpy arrays (x, y coordinate pairs)
        polygons = result.masks.xy
        for poly in polygons:
            print(f"Mask Polygon points: {len(poly)} points")
```
